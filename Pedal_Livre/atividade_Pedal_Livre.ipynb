```python
from sklearn.linear_model import LinearRegression

# Duração do aluguel e o preços
duracao = [[10],[15],[20],[25],[30],[35],[40],[50],[55],[60]]
precos = [10, 13, 14, 18, 19, 23, 24, 30, 32, 34]

# Prepare o modelo
modelo = LinearRegression()

# Treinando o modelo.
modelo.fit(duracao, precos)

# Previsão principal
duracao_previsao = 45

# Previsão do preço
preco_previsao = modelo.predict([[duracao_previsao]])

print(f"Para um aluguel de {duracao_previsao} minutos, o preço previsto é: R$ {preco_previsao[0]:.2f}")

# Outras durações
print(F"Preço previsto para 20 minutos: R$ {modelo.predict([[20]])[0]}:.2f")
print(F"Preço previsto para 60 minutos: R$ {modelo.predict([[60]])[0]}:.2f")
```

7- O modelo estimou R$27,04 para 45 minutos, usando os preços antigos para calcular o valor aproximado do aluguel.
